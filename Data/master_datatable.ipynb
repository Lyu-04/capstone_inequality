```python
''' 
the function below takes a municipality and an outcome variable as input and returns the top predictors for that municipality
based on how abnormal they are compared to the national average and how strongly they correlate with the outcome variable nationwide. 
It also visualises the results in two bar charts: one for the combined score (z-score × nationwide correlation) and one for the z-score alone.
AGAIN this is not part of the project but some potentially useful code for later
'''

def highest_predictors(municipality, variables, outcome_variable, top_n=5):
    


    available_vars = [var for var in variables if var in master_df.columns]
    
    # Get the municipality's value for the outcome
    muni_row = master_df[master_df['name'] == municipality].iloc[0]
    outcome_value = muni_row[outcome_variable]
    




    print(f"📍 {municipality} — {outcome_variable}: {outcome_value:.3f}")
    print(f"   NL average: {master_df[outcome_variable].mean():.3f}\n")
    
    # For each variable, measure how abnormal this municipality is
    # compared to all others (z-score distance)
    results = {}
    for var in available_vars:
        if var == outcome_variable:
            continue
        
        col = master_df[var]
        muni_val = muni_row[var]
        woll
        if pd.isna(muni_val) or col.std() == 0:
            continue
        
        # How many standard deviations away from the mean is this municipality?
        z_score = (muni_val - col.mean()) / col.std()
        
        # Also get the nationwide correlation for context
        nationwide_corr = col.corr(master_df[outcome_variable])
        
        results[var] = {
            'muni_value':        muni_val,
            'nl_average':        col.mean(),
            'z_score':           z_score,          # how abnormal is this municipality
            'nationwide_corr':   nationwide_corr,  # does this variable matter nationwide
            # High z_score + high correlation = strong candidate for root cause
            'combined_score':    abs(z_score) * abs(nationwide_corr)
        }
    
    results_df = pd.DataFrame(results).T.sort_values('combined_score', ascending=False)
    
    # Visualise
    top_df = results_df.head(top_n)
    
    fig, axes = plt.subplots(1, 2, figsize=(16, 5))
    
    # Left: combined score bar chart
    top_df['combined_score'].plot(kind='barh', ax=axes[0], color='steelblue')
    axes[0].set_title(f'Top predictors for {municipality}\n(z-score × nationwide correlation)')
    axes[0].invert_yaxis()
    
    # Right: how far this municipality deviates from NL average on each predictor
    top_df['z_score'].plot(kind='barh', ax=axes[1], color=[
        'tomato' if z > 0 else 'steelblue' for z in top_df['z_score']
    ])
    axes[1].axvline(0, color='black', linewidth=0.8)
    axes[1].set_title(f'{municipality} deviation from NL average\n(standard deviations)')
    axes[1].invert_yaxis()
    
    plt.suptitle(f'Outcome: {outcome_variable}', fontweight='bold')
    plt.tight_layout()
    plt.show()
    
    print(results_df.head(top_n)[['muni_value', 'nl_average', 'z_score', 'nationwide_corr', 'combined_score']].round(3))
    
    return results_df.head(top_n).index.tolist()


# example usage for Diemen and the outcome variable 'c30_wealth_income'
highest_predictors('Diemen', master_df.columns[1:], 'c30_wealth_income', top_n=5)
    
```

```python
# ok now per municipality, show which variable deviates the most froom the national mean
municipality_sailiant_var = pd.DataFrame(columns=['name', 'most_deviant_variables', 'z_scores'])

def plot_top_municipal_deviations():
    # calculate the z-score for each variable in the master_df
    for var in interest_variables:
        col = master_df[var]
        mean = col.mean()
        std = col.std()
        master_df[var +'_z'] = (col - mean) / std

    # for each municipality, find the variable with the top 3 highest absolute z-score
    for idx, row in master_df.iterrows():
        muni_name = row['name']
        z_scores = {var: abs(row[var+'_z']) for var in interest_variables}
        top_vars = sorted(z_scores, key=z_scores.get, reverse=True)[:3]
        top_z_scores = [z_scores[var] for var in top_vars]
       
        municipality_sailiant_var.loc[len(municipality_sailiant_var)] = [muni_name, top_vars, top_z_scores]

    
       

plot_top_municipal_deviations()
municipality_sailiant_var.head(n=100)
    
    


# create hheatmap of the z-scores for each municipality and variable (red is positive deviation from the mean, blue is negative deviation, white is close to the mean)

def municipality_zscore_heatmap(master_df, interest_variables, municipality_col="name"):
    # Keep only municipality name + variables of interest
    df = master_df[[municipality_col] + interest_variables].copy()

    # Compute z-scores for each variable across municipalities
    zscore_df = df.copy()
    for var in interest_variables:
        mean = df[var].mean()
        std = df[var].std()

        if std == 0 or pd.isna(std):
            zscore_df[var] = np.nan
        else:
            zscore_df[var] = (df[var] - mean) / std

    # Set municipality names as row index
    zscore_df = zscore_df.set_index(municipality_col)

    # Plot heatmap
    plt.figure(figsize=(14, max(8, len(zscore_df) * 0.35)))
    sns.heatmap(
        zscore_df,
        cmap="coolwarm",
        center=0,
        linewidths=0.3,
        linecolor="white",
        cbar_kws={"label": "Z-score"}
    )

    plt.title("Municipality × Variable Heatmap of Z-scores")
    plt.xlabel("Variables")
    plt.ylabel("Municipality")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    return zscore_df

d_f = municipality_zscore_heatmap(master_df, interest_variables)
```

```python

def top_3(row):
        row = row['all_all_all_all':]  # skip non-numeric cols
        row = row.replace('  NA', np.nan)
        row = row.dropna()
     
        
        top_vars = row.sort_values(ascending=False).head(3)
        return top_vars.index.tolist(), top_vars.values.tolist()

#the function that takes the dataset and returns a top 3 higest scores for each neighborhoods
def neighborhood_ranking(df, var):

    #check for the NaN values for the dataset
    df = df.replace('  NA', np.nan)

    missing = df.isnull().sum().sort_values(ascending=False)
    missing_percent = (missing / len(df) * 100).round(1)

    summary = pd.DataFrame({
        'missing_count': missing,
        'missing_pct': missing_percent
    }).query('missing_count > 0')

    
    
    #make sure to count out the non-numeric columns gementeenaam
    for neighborhoods in df['name'].unique():
        neighborhood_data = df[df['name'] == neighborhoods]
        top_3_vars, top_3_values = top_3(neighborhood_data.iloc[0])
        print(f"variable: {var}")
        print(f"Neighborhood: {neighborhoods}")
        print(f"Top 3 variables: {top_3_vars}")
        print(f"Top 3 values: {top_3_values}")
        print("\n")
   
    
    return summary, top_3_vars, top_3_values
    
    

```

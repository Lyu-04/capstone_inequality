## Part 1: HTML Basics (5 mins)

### What is HTML?
HTML = **HyperText Markup Language**
- It's the **skeleton** of a web page
- CSS is the **skin/styling**
- JavaScript is the **brain/interactivity**

### HTML Structure: Tags
HTML uses **tags** to tell the browser what content is what:

```html
<tagname>Content goes here</tagname>
```

**Opening tag** + **Content** + **Closing tag** = **Element**

### Example:

```html
<!-- A paragraph -->
<p>This is a paragraph</p>

<!-- A heading -->
<h1>This is a title</h1>

<!-- A link -->
<a href="https://google.com">Click me</a>

<!-- A button -->
<button>Click me</button>
```

### Self-Closing Tags
Some tags don't have content and close themselves:

```html
<img src="photo.jpg" alt="A photo">
<input type="text">
<br>  <!-- Line break -->
<hr>  <!-- Horizontal line -->
```

### Attributes
Tags can have **attributes** that provide extra info:

```html
<a href="page.html">Link</a>
   ^^^^  ^^^^^^^^^^
   attribute name and value

<button class="my-button">Click</button>
         ^^^^^
         attribute name
```

## Part 2: Common HTML Tags (5 mins)

### Text Tags
```html
<h1>Main Heading</h1>          <!-- Largest heading -->
<h2>Subheading</h2>            <!-- Smaller heading -->
<h3>Even smaller</h3>          <!-- And so on... -->
<p>A paragraph of text</p>     <!-- Normal text -->
<strong>Bold text</strong>     <!-- Important text -->
<em>Italic text</em>           <!-- Emphasized text -->
```

### Container Tags
```html
<div>Group content together</div>      <!-- Generic container -->
<section>A section of content</section> <!-- Semantic container -->
<nav>Navigation links</nav>            <!-- For navigation -->
<header>Page header</header>           <!-- Top of page -->
<footer>Page footer</footer>           <!-- Bottom of page -->
```

### Interactive Tags
```html
<a href="url.html">Click to go to page</a>  <!-- Link -->
<button>Click me</button>                      <!-- Button -->
<input type="text">                          <!-- Text input -->
<input type="checkbox">                      <!-- Checkbox -->
<input type="range" min="0" max="100">    <!-- Slider -->
<label>Label for input</label>                <!-- Label text -->
```

## Part 3: Attributes - Making Tags Do Things (5 mins)

### Essential Attributes

| Attribute | Used On | Purpose | Example |
|-----------|---------|---------|----------|
| `id` | Any tag | Unique identifier (use once per page) | `<div id="main-panel">` |
| `class` | Any tag | Style group (can use many times) | `<button class="btn-primary">` |
| `href` | `<a>` | Where link goes | `<a href="page.html">` |
| `type` | `<input>` | Type of input | `<input type="text">` |
| `onclick` | Any tag | Run code when clicked | `<button onclick="doSomething()">` |
| `value` | `<input>` | Default value | `<input value="Hello">` |
| `src` | `<img>` | Image file path | `<img src="photo.jpg">` |
| `alt` | `<img>` | Text if image fails | `<img alt="A photo">` |
| `name` | `<input>` | Name for form data | `<input name="username">` |
| `id` | `<input>` | Connect to label | `<label for="age"> <input id="age">` |

### Examples from Your Code

```html
<!-- From your landing page -->
<a href="analysis.html" class="cta-button">Start Analysis</a>
   ^^^^^^^^^^^^^^^^^        ^^^^^^^^^^^
   Goes to analysis.html    Styled with CSS class

<!-- From your questionnaire -->
<button class="option-button" onclick="selectOption(1, 'Amsterdam Noord')">
        ^^^^^^^^^^^^^^^^                ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
        CSS styling                     JavaScript function called on click

<!-- From your questionnaire (slider) -->
<input type="range" min="0" max="80000" value="40000" id="incomeSlider" oninput="updateIncomeDisplay()">
       ^^^^^^^^^                                          ^^^^^^^^^^^^^^^  ^^^^^^^^^^^^^^^^^^^^^^^
       Slider type                                        Unique ID        Called while dragging

<!-- From your report page -->
<div id="summaryGrid"></div>
     ^^^^^^^^^^^
     Unique ID so JavaScript can find this element
```

## Part 4: The Head Section (5 mins)

Every HTML page starts with this structure:

```html
<!DOCTYPE html>
<html lang="en">
<head>
  <!-- Metadata goes here -->
</head>
<body>
  <!-- Visible content goes here -->
</body>
</html>
```

### What Goes in the `<head>`?

```html
<head>
  <meta charset="UTF-8">                    <!-- Text encoding -->
  <meta name="viewport" content="width=device-width, initial-scale=1.0">  <!-- Mobile support -->
  <title>Page Title</title>                  <!-- Browser tab title -->
  
  <!-- Link CSS file -->
  <link rel="stylesheet" href="style/style.css">
  
  <!-- Link Google Fonts -->
  <link href="https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600&display=swap" rel="stylesheet">
  
  <!-- Include JavaScript library -->
  <script src="https://some-library.com/lib.js"></script>
</head>
```

### What Goes in the `<body>`?
Everything the user **sees**: text, buttons, images, etc.

## Part 5: Deconstructing Your Code (10 mins)

Let's break down your **landing page** (`index.html`) piece by piece:

### The Full Structure

```html
<!DOCTYPE html>                           <!-- Tell browser: this is HTML5 -->
<html lang="en">                         <!-- Language: English -->
<head>
  <meta charset="UTF-8">                 <!-- Text encoding -->
  <meta name="viewport" content="width=device-width, initial-scale=1.0">  <!-- Mobile responsive -->
  <title>Amsterdam Inequality Analysis - Home</title>  <!-- Browser tab title -->
  
  <!-- Import font from Google -->
  <link href="https://fonts.googleapis.com/css2?family=Inter..." rel="stylesheet">
  
  <!-- Link to your CSS file -->
  <link rel="stylesheet" href="style/style.css">
  
  <!-- CSS code directly in HTML (for this page) -->
  <style>
    /* CSS rules go here */
  </style>
</head>
<body>
  <!-- Navigation bar -->
  <nav class="navbar">
    <div class="navbar-brand">
      <strong>Inequality Explorer</strong>
    </div>
    <div>
      <a href="index.html">Home</a>              <!-- Link to this page -->
      <a href="i_map.html">Overview</a>         <!-- Link to map page -->
    </div>
  </nav>

  <!-- Main content -->
  <div class="landing-container">
    <div class="hero">
      <h1>Welcome to the Inequality Analysis Tool</h1>
      <p>Placeholder text: Explore inequality patterns...</p>
    </div>

    <!-- Button that goes to questionnaire -->
    <a href="analysis.html" class="cta-button">Start Analysis</a>
  </div>

  <!-- Footer -->
  <footer class="footer">
    <p>© 2026 Amsterdam Inequality Explorer</p>
  </footer>
</body>
</html>
```

### Breaking It Down:

**1. The Navigation**
```html
<nav class="navbar">              <!-- Navigation container -->
  <div class="navbar-brand">
    <strong>Inequality Explorer</strong>  <!-- Bold branding text -->
  </div>
  <div>
    <a href="index.html">Home</a>         <!-- Link to home -->
    <a href="i_map.html">Overview</a>    <!-- Link to map -->
  </div>
</nav>
```

**2. The Hero Section**
```html
<div class="hero">                <!-- Styled container -->
  <h1>Welcome...</h1>              <!-- Main heading -->
  <p>Placeholder text...</p>       <!-- Paragraph -->
</div>
```

**3. The Call-to-Action Button**
```html
<a href="analysis.html" class="cta-button">Start Analysis</a>
   ^^^^^^^^^^^^^^               ^^^^^^^^^^^  
   Points to questionnaire      CSS styling (looks like button)
```
It's actually an `<a>` tag styled to look like a button!

## Part 6: The Questionnaire Page Breakdown (10 mins)

This is more complex because it has **interactivity**. Let's look at key parts:

### Progress Bar

```html
<div class="progress-bar">                    <!-- Container for progress -->
  <div class="progress-fill" id="progressFill" style="width: 0%;"></div>
                              ^^^^^^^^^^^^^^^                 ^^^^^^^^^^^^^^^  
                              Unique ID so JS can find it    JavaScript updates this
</div>
<div class="progress-text" id="progressText">Question 1 of 4</div>
                             ^^^^^^^^^^^^^^^^
                             JavaScript updates this too
```

### Question Container

```html
<!-- Question 1 - starts visible (active) -->
<div class="question-step active" id="step-1">  <!-- 'active' = shown -->
  <div class="question-title">Which neighbourhood do you want to see?</div>
  
  <!-- Button options -->
  <div class="options-container">
    <button class="option-button" onclick="selectOption(1, 'Amsterdam Noord')">
                                    ^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^^
                                    Calls JavaScript when clicked
      Amsterdam Noord
    </button>
    <!-- More buttons... -->
  </div>
  
  <!-- Error message (hidden by default) -->
  <div class="error-message" id="error-1">Please select a neighbourhood</div>
                               ^^^^^^^^
                               JavaScript shows this if user forgets to select
</div>

<!-- Questions 2, 3, 4 - hidden by default (no 'active' class) -->
<div class="question-step" id="step-2">  <!-- 'display: none' in CSS by default -->
  <!-- Question 2 content -->
</div>
```

### Multi-Select (Question 3: Domains)

```html
<div class="question-step" id="step-3">
  <div class="question-title">Select domain(s)</div>
  
  <div class="checkbox-container">
    <!-- Each checkbox can be selected independently -->
    <label class="checkbox-item">
      <input type="checkbox" value="Health" onchange="selectDomain('Health')">
                              ^^^^^^^^                ^
                              Checkbox type           Called when checked/unchecked
      <span>Health</span>
    </label>
    
    <label class="checkbox-item">
      <input type="checkbox" value="Income" onchange="selectDomain('Income')">
      <span>Income</span>
    </label>
    <!-- More checkboxes... -->
  </div>
</div>
```

### Slider (Question 4: Income Group)

```html
<div class="question-step" id="step-4">
  <div class="question-title">Select income group</div>
  
  <div class="slider-container">
    <input type="range" 
           min="0" 
           max="80000" 
           value="40000" 
           id="incomeSlider" 
           oninput="updateIncomeDisplay()">
                    ^
                    Called while dragging (updates in real-time)
    
    <!-- Shows the value (e.g., "€40,000") -->
    <div class="slider-value" id="incomeValue">€40,000</div>
                               ^^^^^^^^^^^^^
                               JavaScript updates this
  </div>
</div>
```

### Navigation Buttons

```html
<div class="button-group">
  <!-- Back button (hidden on first question) -->
  <button class="btn btn-secondary" 
          id="backBtn" 
          onclick="previousStep()" 
          style="display: none;">  <!-- Hidden by default -->
    ← Back
  </button>
  
  <!-- Next button (changes text on final question) -->
  <button class="btn btn-primary" 
          id="nextBtn" 
          onclick="nextStep()">
    Next →
  </button>
</div>
```

## Part 7: How JavaScript Connects to HTML (5 mins)

### The Connection: IDs and Classes

JavaScript uses **IDs** and **classes** to find and modify HTML:

```html
<!-- HTML -->
<button id="nextBtn" onclick="nextStep()">Next</button>
        ^^^^^^^^^^^             ^^^^^^^^
        Unique ID               Function to call
```

```javascript
// JavaScript
document.getElementById('nextBtn').textContent = 'Generate Report';
// Finds the button by ID and changes its text
```

### onclick vs oninput

```html
<!-- Called ONCE when clicked -->
<button onclick="nextStep()">Next</button>

<!-- Called CONTINUOUSLY while dragging -->
<input type="range" oninput="updateIncomeDisplay()">
```

## Part 9: Quick Reference Cheat Sheet (5 mins)

### Essential Tags

```html
<!-- Structure -->
<!DOCTYPE html>
<html> </html>
<head> </head>
<body> </body>

<!-- Text -->
<h1>Heading 1</h1> ... <h6>Heading 6</h6>
<p>Paragraph</p>
<strong>Bold</strong>
<em>Italic</em>

<!-- Containers -->
<div>Generic container</div>
<nav>Navigation</nav>
<header>Header</header>
<footer>Footer</footer>
<section>Section</section>

<!-- Links & Buttons -->
<a href="page.html">Link</a>
<button onclick="function()">Click Me</button>

<!-- Forms -->
<input type="text">              <!-- Text input -->
<input type="checkbox">          <!-- Checkbox -->
<input type="range" min="0" max="100">  <!-- Slider -->
<label>Label</label>

<!-- Images -->
<img src="photo.jpg" alt="Description">
```

### Essential Attributes

```html
id="uniqueName"              <!-- Unique identifier -->
class="styleName"            <!-- CSS styling -->
href="page.html"             <!-- Link destination -->
onclick="function()"         <!-- Run function when clicked -->
onchange="function()"        <!-- Run function when changed -->
oninput="function()"         <!-- Run function while changing -->
type="text/checkbox/range"   <!-- Input type -->
value="defaultValue"         <!-- Default value -->
style="color: red;"          <!-- Inline CSS -->
disabled                      <!-- Disable element -->
```

### Page Navigation Pattern

```html
<!-- Navbar on every page -->
<nav class="navbar">
  <a href="index.html">Home</a>
  <a href="analysis.html">Analysis</a>
  <a href="i_map.html">Overview</a>
</nav>

<!-- Content -->
<div class="container">
  <!-- Page content -->
</div>

<!-- Footer on every page -->
<footer>
  <p>© 2026</p>
</footer>
```
